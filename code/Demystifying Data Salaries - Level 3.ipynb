## 💾 The Data

The data, available in `salaries.csv`, comes from a survey hosted by the aforementioned HR consultancy. Each data point represents a single employee's salary record for a given year.

| Column    | Description                                                      |
|-----------|------------------------------------------------------------------|
| `work_year`  | The year the salary was paid. |
| `experience_level`  | Employee experience level. `EN`: Entry-level / Junior, `MI`: Mid-level / Intermediate, `SE`: Senior / Expert, `EX`: Executive / Director |
| `employment_type`  | Employment type. `PT`: Part-time, `FT`: Full-time, `CT`: Contract, `FL`: Freelance |
| `job_title`  | The job title during the year. |
| `salary`  | Gross salary paid (in local currency). |
| `salary_currency`  | Salary currency (ISO 4217 code). |
| `salary_in_usd`  | Salary converted to USD using average yearly FX rate. |
| `employee_residence`  | Employee's primary country of residence (ISO 3166 code). |
| `remote_ratio`  | Percentage of remote work. `0`: No remote work (<20%), `50`: Hybrid (50%), `100`: Fully remote (>80%) |
| `company_location`  | Employer's main office location (ISO 3166 code). |
| `company_size`  | Company size. `S`: Small (<50 employees), `M`: Medium (50–250 employees), `L`: Large (>250 employees) |

This section utilized regression techniques to predict employee salaries using information pertaining to their experience level, remote ratio, & company location. In particular, **Analysis II** featured linear & random forest regression. Some hyperparameter tuning was performed on the random forest regression model to help try to elevate its performance, & so three unique models were produced. R-squared scores & error metrics were used to evaluate these models' predictive capabilities.

Regarding feature importances of the models, there was one common finding across all three models--employees' experience level is considerably important when it comes to predicting their salaries. This finding reinforces conclusions obtained in **Analysis I** in which salaries were observed to generally increase with greater experience.  
In regard to the other two feature variables--remote ratio & company location--these two methods came to opposing conclusions. The linear regression model determined that an employee's company location is somewhat important in predicting their salary, & that the amount of remote work they do is of very little to no importance. On the other hand, both of the random forest models found that the amount of remote work an employee does is somewhat important in predicting their salary, whereas their company location is not really important.

As such, the only meaningful conclusion that can be made from this analysis is that an employee's experience level is considerably important when it comes to predicting their salary.

### Experience Level
When it comes to employees' experience, it would be logical to hypothesize that more experience leads to greater knowledge, greater aptitude, a more well-rounded skillset, & more confidence in performing the roles & responsibilities of a job. Furthermore, various jobs require various levels of expertise, implying that certain jobs will be naturally be biased & restricted towards more experienced candidates. Generally speaking, both of these factors will naturally lead to more challenging jobs offering higher salaries.  
In this case, data scientists typically possess more well-rounded skillsets & deeper understandings of statistical concepts & analytical techniques & algorithms that data analysts. To achieve such levels of understanding often requires more experience. As such, companies hiring data scientists would typically be looking for more experienced individuals than they would be for data analysts. So far, it has been observed that data scientists typically make more than data analysts. Therefore, it would be reasonable to hypothesize that more experienced individuals typically make more than less experienced employees.

In this dataset, experience level is grouped into four categories which are listed below. Given that more experience in the tech industry typically requires more knowledge & stronger skillsets, such experience levels are likely to be hierarchical in some way. Most people start at the bottom then work their way up throughout their career as they encounter more experiences & develop more skills. As salaries are explored across this variable, it can be useful to keep this in mind.
- **EN:** Entry-level / Junior. There are 2,804 corresponding survey responses.
- **MI:** Mid-level / Intermediate. There are 5,225 corresponding survey responses.
- **SE:** Senior / Expert. There are 12,455 corresponding survey responses.
- **EX:** Executive / Director. There are 269 corresponding survey responses.

Ultimately, similar analyses can be done here as those for employees' country of residence. Defining the 'experience_level' variable as ordinal can help extend this sense of hierarchy between the four experience levels.

```python
#Make the 'experience_level' variable ordinal
df_I_jobs['experience_level'] = pd.Categorical(df_I_jobs['experience_level'], ordered=True, categories=["EN","MI","SE","EX"])

#Inspect the 'experience_level' values
#display(df_I_jobs['experience_level'].value_counts())

#Inspect the salaries of these positions across their 'experience_level'
df_I_exp_sals = df_I_jobs.groupby('experience_level')['salary_in_usd'].describe().reset_index()
df_I_exp_sals.iloc[:,1:] = round(df_I_exp_sals.iloc[:,1:], 2)
df_I_exp_sals['count'] = df_I_exp_sals['count'].astype('int')
display(df_I_exp_sals.sort_values('mean', ascending=False))

#Inspect the salaries by position & experience level
df_I_jobs_exp_sals = df_I_jobs.groupby(['experience_level','job_title'])['salary_in_usd'].describe().reset_index()
df_I_jobs_exp_sals.iloc[:,3:] = round(df_I_jobs_exp_sals.iloc[:,3:], 2)
df_I_jobs_exp_sals['count'] = df_I_jobs_exp_sals['count'].astype('int')
display(df_I_jobs_exp_sals)
```

# Analysis II - Predictive Modeling
**_Develop a predictive model to estimate an employee’s salary (in USD) using experience level, company location, and remote ratio. Which features are the strongest predictors of salary?_**

To further explore relationships between the various variables & peoples' salaries, machine learning models can be used. By incorporating factors into the predictive process, connections can be evaluated to see how variables change in relation to one another, particularly the salaries. Given that the target variable is numerical, this problem calls for a regression model, three types of which are described below.
- Linear regression is the most generic form & is usually appropriate when variables exhibit linear relationships.
- Lasso regression is similar to linear regression but it incorporates regularization which essentially penalizes more extreme data points such that models avoid overfitting--a phenomena in which models perform strongly on training data but poorly on new, unseen data.
- Random forest regression can handle more complex data, specifically when it does not possess linear relationships. This method may be more effective in this case given the number of variables being incorporated.

Building multiple models can provide an overall assessment as to relationships between variables as well as the effectiveness & capabilities of different algorithms. In this case, linear & random forest regression techniques were utilized.  
Since two of the three feature variables are non-numerical, they need to be converted to numerical values for `scikit-learn` functions to process them. `pandas.get_dummies()` can be used to do so which creates a new column for each categorical value with 1's & 0's indicating whether a data point contains a specific value or not. It is prudent to use this technique in this case given that the feature variables being utilized are nominal--there is no inherent order in their values.  
Moreover, hyperparameter tuning can be utilized for the random forest models to find more optimal model parameters that can help improve predictive capabilities. For these models, some relevant parameters are listed below. A grid search cross validation technique can be used to assess the model with different hyperparameter values such that the most optimal values can be found.
- 'n_estimators': The number of trees in a forest. More trees can improve model performance at the cost of computational resources.
- 'max_depth': The max depth of each tree. Deeper trees can model more complex relationships at the cost of overfitting.
- 'min_samples_split': The minimum number of samples required to split an internal node. Higher values can help to prevent overfitting.
- 'min_samples_leaf': The minimum number of samples required to be at a leaf node. Higher values make the model more conservative.
- 'max_features': The number of features to consider when looking for the best split.

When evaluating the performance of a regression model, error terms & similar metrics are often used. In particular, the root-mean-square-error quantity quantifies the average difference between predicted & actual data points. The R-squared metric can provide an assessment as to how well a model explains the variability in the target variable. In other words, it evaluates how accurate predictions of a model are.

To investigate the feature variables of a predictive model & how strong they are, feature analysis can be performed which quantifies how much a feature variable contributes to a model's predictions. Ultimately, the specifics as to how these quantities are measured depends on the model, but feature importances generally give an indication as to how relevant a variable is in the prediction process. It's important to note that these quantities don't indicate the strength of the relationship between the variables.


To begin, the non-numerical feature variables need to be converted to numerical data, the feature & target data needs to be clearly defined & then segmented into training & testing data. The latter will be used to obtain an assessment as to the accuracy of the model. In this case, the test size will be set to 25% of the dataset.  
Additionally, correlation can be calculated between the remote-ratio & salary variables to get an initial idea as to the relationship between how much people work remotely & their salary.

```python
#Calculate correlation between 'remote_ratio' & salary
sal_rem_corr = salaries_df[['remote_ratio','salary_in_usd']].corr().iloc[0,1]
display(f"The correlation between how much people work remotely & their salaries is about: {round(sal_rem_corr, 3)}.")

#Filter for the relevant data
df_II = salaries_df[['experience_level', 'company_location', 'remote_ratio', 'salary_in_usd']].copy()

#Encode dummy variables for categorical variables
df_II_dummies = pd.get_dummies(df_II, drop_first=True)
display(df_II_dummies.head(2))

#Assign the feature, target data
X = df_II_dummies.drop(columns=['salary_in_usd'])
y = df_II_dummies['salary_in_usd']

#Split the data into training/testing sets
X_train, X_test, y_train, y_test = train_test_split(X, y, random_state=10, test_size=0.25)
```

### Analysis II - Linear Regression
The `LinearRegression` function from scikit-learn can be used to create such a model. To evaluate the model, the `mean_squared_error` & `.score` functions can be used which retrieve the mean-squared-error & r-squared scores respectively. Feature importances can be retrieved using the `.coef_` attribute.
- For regression models, the `.score()` function retrieves the same r-squared quantity as the `r2_score` function so either could be used.

```python
#Instantiate a linear regression model
LR_model = LinearRegression()

#Fit the model to the data
LR_model.fit(X_train, y_train)

#Cast predictions
LR_preds = LR_model.predict(X_test)

#Retrieve the model's MSE, R2-score
display(mean_squared_error(y_test, LR_preds, squared=False))
    # ~ 65,558
display(f"R-squared coefficient: {LR_model.score(X_test, y_test)}.")
#display(r2_score(y_test, LR_preds))
    #same as .score(); ~ 0.164

#Retrieve the feature importances from the model
LR_feat_coefs = LR_model.coef_

#Reorganize the feature importance data
LR_feat_imp = pd.DataFrame(LR_feat_coefs, index=X.columns, columns=["Coefficients"]).reset_index().rename(
                        columns={'index':'Variable'})
display(LR_feat_imp.head(3))
```

This linear regression model--built on experience level, company location, & remote ratio--has an r-squared score of about 16.4%. In other words, these variables explain about 16% of the variance in employee salaries which is a relatively small quantity. Furthermore, the mean-square-error term of 65,500 is also quite substantial. As such, these measures seem to indicate that linear regression is not a good fit for the data & predicting salaries.

To examine the importances of the feature variables, averages can be compiled across the subgroups of the three relevant feature variables. Recall that the 'experience_level' & 'company_location' variables are categorical. To get an overall assessment as to the importance of these variables, averages can be calculated the subgroups.

```python
#Want to retrieve the avg feature importance coef's across the subgroups of 'experience_level', 'company_location'
#Specifically want to get the average magnitude, so need to make sure to retrieve the absolute values of the coef's
LR_FI_abs = LR_feat_imp.copy()
LR_FI_abs['Coefficients'] = LR_FI_abs['Coefficients'].apply(lambda x: np.abs(x))

#Calculate the averages for these 2 variables
LR_FI_avgs_dict = {'experience_level_avg':[LR_FI_abs[LR_FI_abs['Variable'].str.contains('experience')].Coefficients.mean()],
               'company_location_avg':[LR_FI_abs[LR_FI_abs['Variable'].str.contains('company')].Coefficients.mean()]}

#Compile the aggregate feature importances for each feature variable
LR_FI_agg = pd.concat([LR_FI_abs.iloc[:1,:], pd.DataFrame(LR_FI_avgs_dict).T.reset_index().rename(
            columns={'index':'Variable',0:'Coefficients'})], axis=0).reset_index(drop=True)
display(LR_FI_agg)
```

```python
#Visualize the aggreate feature importances of the 3 feature variables
ax1 = sns.barplot(data=LR_FI_agg.sort_values('Coefficients', ascending=False), x='Coefficients', y='Variable')
#edit plot details
plt.title('Feature Variable Importance - Linear Regression')
plt.xlabel('Magnitude of Importance')
plt.xticks(fontsize=9)
plt.ylabel('Feature Variable')
ax1.set_yticklabels(labels=['experience_level', 'company_location', 'remote_ratio'], fontsize=9)
plt.grid(axis='x',alpha=0.1,color='gray')
plt.show();
```

The plot above displays the average quantitative importances of each of the three feature variables using to predict employee salaries with a linear regression model. This model was built using over 57,000 data points. The 'experience_level' & 'company_location' variables have categorical values, so the model used quantitative values to predict salaries. As such, feature importances were retrieved for each subcategory. These values were then averaged & compiled into a single metric to obtain a more general assessment, the results of which are included in the plot above.  
Generally speaking, the larger the importance, the more important the feature in predicting employee salaries. It's important to note that these values don't indicate the strength of a relationship like correlation does.

Based on this data, employees' experience level is very important in predicting their salaries when using linear regression whereas their remote ratio is not. This is generally consistent with findings from previous analyses in that salary increases with experience & that salary is generally the same for people who work mostly remotely (100) & mostly in-person (0). Meanwhile, the location of employees' company is somewhat important in predicting their salaries.

### Analysis II - Random Forest Regression
To try & explore the relationships between these variables at a deeper level, random forest models can be used. A random forest model is composed of multiple decision tree models. More specifically, a tree is an individual decision tree model that is trained on a slightly different subset of data than all other trees. Final predictions are made by compiling predictions across all trees.

One primary hyperparameter for random forest regressor models is the "n_estimators" parameter, which specifies the number of trees to use in a random forest model. This hyperparameter was tuned in isolation to evaluate how it affected the model's performance. Later, this parameter was also tuned in combination with other parameters using grid search cross validation.

In regards to feature importance, it can essentially be described as how much this feature is used in each tree of the forest. So, larger feature importance coefficients indicate that a feature variable is more important in predicting the target variable in a random forest model.

```python
#Assess different 'n_estimators' values of a RF model on the training data. Make sure to set the random state
scores = []
for n in [10, 20, 50, 100, 200]:
    #instantiate the model with different # trees
    rf = RandomForestRegressor(n_estimators=n, random_state=40)
    #fit the model to the training data
    rf.fit(X_train, y_train)
    #compile the R^2 scores for each model
    scores.append(rf.score(X_test, y_test))

display(scores)
```

For this random seed, the most optimal number of trees for this data is 20 because the corresponding model produces the largest R-squared score (~ 0.1662). A r-squared score of 16% is rather poor & indicates that the model is underperforming. As such, exploring more hyperparameters may be helpful.
