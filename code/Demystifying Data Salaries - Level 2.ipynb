## 💾 The Data

The data, available in `salaries.csv`, comes from a survey hosted by the aforementioned HR consultancy. Each data point represents a single employee's salary record for a given year.

| Column    | Description                                                      |
|-----------|------------------------------------------------------------------|
| `work_year`  | The year the salary was paid. |
| `experience_level`  | Employee experience level. `EN`: Entry-level / Junior, `MI`: Mid-level / Intermediate, `SE`: Senior / Expert, `EX`: Executive / Director |
| `employment_type`  | Employment type. `PT`: Part-time, `FT`: Full-time, `CT`: Contract, `FL`: Freelance |
| `job_title`  | The job title during the year. |
| `salary`  | Gross salary paid (in local currency). |
| `salary_currency`  | Salary currency (ISO 4217 code). |
| `salary_in_usd`  | Salary converted to USD using average yearly FX rate. |
| `employee_residence`  | Employee's primary country of residence (ISO 3166 code). |
| `remote_ratio`  | Percentage of remote work. `0`: No remote work (<20%), `50`: Hybrid (50%), `100`: Fully remote (>80%) |
| `company_location`  | Employer's main office location (ISO 3166 code). |
| `company_size`  | Company size. `S`: Small (<50 employees), `M`: Medium (50–250 employees), `L`: Large (>250 employees) |

# Summary of Findings & Recommendations
This project (part two of three) analyzed over 57,000 survey responses of individuals within the tech industry regarding their employment details between 2020-2024. As no preprocessing was necessary, the analyses examined specific subsets of individuals & their salaries. In particular, employment & salary information was evaluated based on the top-earning positions, how much employees worked remotely, & the size of their company. In assessing such individuals, a deeper understanding as to some recent/current trends of employees & salaries within the tech industry was uncovered. ### Analysis I - Highest-Paying Jobs
Of the 57,194 survey responses, the jobs with the highest average salaries were identified to get a basic understanding as to which positions saw the most pay. To account for less-than-representative job titles, two methods were employed to obtain these job titles.

Firstly, the five jobs with the greatest average salaries across all of the responses were identified & examined. They included (from highest to lowest): Analytics Engineering Manager, Data Science Tech Lead, Applied AI ML Lead, Head of Machine Learning, & Engineering Manager. Their average salaries ranged between \\$262,000-\\$400,000. The first three titles had either one or two records each & so it is difficult to weigh the reliability of the associated observations. The latter jobs two had nine & 268 data points respectively, so they are relatively more representative.

Second, the five jobs with the greatest average salaries of those with at least three survey responses were retrieved & analyzed. They included (from highest to lowest): Head of Machine Learning, Engineering Manager, Head of AI, Machine Learning Model Engineer, & Director of Machine Learning. Their average salaries ranged between \\$242,000-\\$289,000. While the Head of Machine Learning sees the highest average salary of these jobs, they also see the smallest outright salary & the most volatility. As such, estimating salaries of Machine Learning Heads may be more challenging. The job with the most consistent salaries of these five positions is the Machine Learning Model Engineer. Overall, the more consistent the salaries of these positions, the more reliable & predictable they are for companies who may be looking to hire such positions.

Across both of these methods, there were some commonalities in the positions. Firstly, two of the jobs--Head of Machine Learning & Engineering Manager--directly overlapped. Additionally, of these eight positions, seven of them refer to titles that directly oversee & direct other employees such as managers, leads, heads, or directors. Moreover, three of them relate to engineering, four relate to machine learning (ML), & two relate to artificial intelligence (AI).

As such, managerial jobs across these three fields appear to be particularly specialized & therefore command higher levels of pay within the tech industry. Such jobs are likely to be very competitive because companies will want to make sure that candidates are fully capable & worthy of positions that involve many responsibilities. As a consequence, companies should be willing to offer strong salaries to candidates whom meet their satisfaction in the hiring process.


### Analysis II - Remote Work & Average Salary
With continuing developments in technology, the ability for people to work fully remotely & remain effective has become more accessible. Although fully-remote work may impact work within the office building of a company or business, such as ease of communication & working environment, this style of work can also improve accessibility for both companies & employees in terms of who they can hire & where they can work respectively. In hiring employees who work fully remotely, companies can have a wider range of candidates whom they can choose from to fill positions. For hybrid work, employees still need to be near their office & so hiring such candidates will still be as restricted as for those who work fully in person.  
Ultimately, whether companies prefer fully-remote or in-person work will depend on various factors, but hybrid work will likely require additional resources & levels of accessibility because companies will have to handle people who work both in the office & outside it. As such, hybrid work may be less desirable for businesses.

Reported salaries were analyzed according to how much remote work employees did which was described using three remote ratios. Employees either worked mostly remotely, rarely worked remotely (in person), or worked evenly between remote work & non-remote work (hybrid). Through analyzing salaries across these remote ratios, trends & differences were obtained that might be indicative as to how popular these three types of work have been & what consequences they might associate with.

Firstly, while there large sample sizes of responses across all three remote ratios, only 271 responses correspond to hybrid employees, whereas the other two ratios each accounted for at least 12,500 responses. Such an imbalance could point to  preferences among companies who hire such individuals & the tech industry in general in that hybrid work is not really offered. Alternatively, it could point to a bias in the people who were surveyed in that hybrid workers were not represented equally.

With this established, salaries were aggregated both overall & on an annual basis. Generally, hybrid workers reported salaries (~ \\$80,000) that were typically about half of those of either in-person or fully remote employees (~ \\$160,000 & \\$150,000 respectively). Such an imbalance may be related to a hiring preferential of companies in that they want candidates to fully commit to one way or the other when it comes to remote work.  
Over the 2020-2024 period, average salaries of the three remote-work types were within about \\$30,000 of each other; however, by 2024, this difference had grown beyond \\$80,000. In particular, average salaries of fully remote & in-person employees grew significantly, whereas that of hybrid employees actually decreased by about \\$5,000.

Given these findings, it is fairly evident that in-person work became increasingly more valuable over these five years. Fully remote work also saw an increase in value but not quite to the same extent. Hybrid work, on the other hand, saw a small decrease in value from 2020-2024. Consequentially, it could be inferred that tech companies generally prioritized in-person & fully remote positions significantly more than hybrid positions. Furthermore, hybrid employees seemingly would generally have had, & might in the near future, more difficulty in obtaining a hybrid position that pays well.


### Analysis III - Company Size & Salary
Reported salaries were analyzed according to the size of the company employees worked at. Companies were labeled as either small (less than 50 employees), medium (between 50-250 employees), or large (more than 250 employees). Through analyzing salaries across these company sizes, trends & differences were obtained that might be indicative as to how capable &/or conservative companies were when it comes to providing competitive salaries.

Although this is slightly logical given how the groups were defined, such an imbalance could give a general indication as to the size of most companies within the tech industry. Alternatively, it could point to a bias in the people who were surveyed in that small & large companies were not represented equally to medium-size companies.

With that aside, salaries were aggregated both overall & on an annual basis. Generally, workers from small companies reported salaries (~ \\$88,000) that were roughly half of those from medium-size or large companies (~ \\$160,000 & \\$153,000 respectively). Such an imbalance may be related to hiring capabilities of companies, their resources, & how they chose to allocate them when it comes to the salaries of their employees.  
Over the 2020-2024 period, average salaries of the three companies were within roughly \\$37,000 of each other, but this disparity grew to over \\$70,000 in 2024. In particular, average salaries of employees from medium-size & large companies grew considerably, whereas those of employees from small companies only increased by about \\$16,000. Moreover, average salaries of those from medium-size & large companies were within \\$14,000 of one another in four of the five years, & neither was consistently ahead of the other.

As such, it is fairly evident that employees in smaller companies faced an increasing disadvantage when it comes to salary in the tech industry during this time period. Across the three company sizes, salaries have generally been fairly consistent & therefore predictable aside from the major spike in the salaries of medium-size companies in 2021. While salaries of employees from medium-size & large companies have been relatively comparable as of recent, those from large companies experienced more stability. The only other significant difference between these two larger categories is that medium-size companies offer much higher salaries on the extreme end (up to \\$800,000) compared to large companies (up to \\$425,000); however, this could be due to a lack of data on employees from large companies.

Consequently, it could be inferred that having more resources, as larger companies tend to have, is generally more of a benefit for employees in terms of what they earn & for companies in terms of what they can offer. As such, it allows them to more easily attract stronger talent.

```python
#import the data & inspect it
salaries_df = pd.read_csv('salaries.csv')
display(salaries_df.head(5))
display(salaries_df.info())
```

# Analysis I - Highest-Paying Jobs
_**Create a bar chart displaying the top 5 job titles with the highest average salary (in USD).**_

By identifying the jobs with the highest average salary, it can provide salary expectations for companies when looking to hire certain positions & how expensive they might be relative to other jobs.

When evaluating these jobs, it would useful to get an idea as to the distribution. If one job title has a very high average salary but there are only a couple of corresponding records, then it may not be very representative. As such, a threshold as to how many corresponding records job titles have may need to be applied when investigating those with the highest average salaries.

```python
#obtain summary statistics of "salary_in_usd" per "job_title"
q1_data_I = salaries_df.groupby('job_title').salary_in_usd.describe().reset_index()
#round salaries to 2 decimals
q1_data_I.iloc[:,2:] = round(q1_data_I.iloc[:,2:], 2)
#convert the "count"s to integers
q1_data_I['count'] = q1_data_I['count'].astype('int')

display(q1_data_I.sort_values('count', ascending=False).reset_index(drop=True).head(5))
#inspect distribution of job frequencies
display(q1_data_I['count'].describe())
```

In the dataset, there are 253 unique job titles. 20 of them only account for one data point, whereas the most popular job--data scientists--accounts for 9,706 records. Most of the jobs that are present account for between three & 50 observations each. This information is prevalent because some jobs with few observations may not possess a sufficient, representative number of data points when evaluating which have the largest average salaries. For instance, if a certain job has three survey responses with reported salaries of \\$1,000,000, \\$250,000, & \\$100,000, the average salary would be \\$450,000; however, these salaries are so inconsistent with one another that they don't provide a reliable, central idea as to what salary such a position might be expected to typically have.  
On the other hand, metrics of jobs with many observations will be more diluted & less prone to influence from extreme values. Without normalized data, calculating aggregate statistics like an average can involve biases.

As an alternative to retrieving the jobs with the greatest average salary, jobs with too few observations could be ignored from this process so those that are more representative can be identified; however, it should be noted that outliers do exist in many real-world datasets & applications. In this case, given the number of records per job in this dataset, a better range would include jobs that have at least three records which is the 25th percentile of counts of "job_title"s. Of the 253 unique jobs, 56 of them have less than three unique survey responses (~ 22%); however, these 56 jobs only account for 92 survey responses.

For simplicity, two processes were utilized.
1) Find the jobs with the top average salaries across the entire dataset.
2) Of jobs with at least two records, find the jobs with the top average salaries across the entire dataset.

```python
#1) Find 5 jobs with highest avg salaries -- All jobs
q1_data_II = q1_data_I.sort_values('mean', ascending=False).reset_index(drop=True)[:5]
display(q1_data_II)

#Plot the data
plt.figure(figsize=(10,5))
ax1 = sns.barplot(data=q1_data_II, x='job_title', y=q1_data_II['mean']/100000)
#label # records per job
ax1.bar_label(ax1.containers[-1], labels=[f"{num} record(s)" for num in q1_data_II['count']], fontsize=8.5)

#edit plot details
plt.title('Jobs with Highest Average Salary - All Jobs')
plt.xlabel('Job')
plt.xticks(fontsize=8, rotation=5)
plt.ylabel('Average Salary ($100,000)')
plt.yticks(fontsize=9)
plt.grid(axis='y',alpha=0.3,color='gray')
plt.show();
```

The plot above displays the average salary of five jobs in the tech industry. These five jobs have the highest average salaries across the whole dataset. The labels above each bar indicate the number of responses each job accounts for in the survey from which the data was collected. Note that the vertical axis is in units of \\$100,000. For reference, the average salary across the whole dataset is about \$159,000 (from _**Demystifying Data Salaries Level 1**_).

The first two jobs, Analytics Engineering Manager & Data Science Tech Lead, have average salaries that are considerably greater than that of any other (by more than \$80,000); however, they only correspond to one survey response each.  
Two of these highly paid jobs involve engineering managers. Two others have to do with machine learning (ML). Five of these jobs are positions that oversee the work of others--managers, leads, heads.

As discussed earlier, jobs with only a handful of observations may potentially pose as outliers. With more data points, extreme salaries are more likely to be diluted & regress to a more meaningful average. As such, the average salaries of the first two jobs--Analytics Engineering Manager & Data Science Tech Lead--aren't as credible or representative because they only have one data point each. Specifically, the significant gap in average salary between these two jobs & the others could be indicative as to the potentially unrepresentative nature of these data points. The last two jobs, head of machine learning & engineering manager, have the most credible data points of these five jobs given that they correspond to nine & 268 records respectively.

```python
#2) Find 5 jobs w/highest avg salaries -- Jobs with 3+ records
q1_data_III = q1_data_I[q1_data_I['count'] > 2].sort_values('mean', ascending=False).reset_index(drop=True)[:5]
display(q1_data_III)

#plot the data
plt.figure(figsize=(10,5))
ax1 = sns.barplot(data=q1_data_III, x='job_title', y=q1_data_III['mean']/100000)
#label # records per job
ax1.bar_label(ax1.containers[-1], labels=[f"{num} record(s)" for num in q1_data_III['count']], fontsize=8.5)

#edit plot details
plt.title('Jobs with Highest Average Salary - Jobs with 3+ Records')
plt.xlabel('Job')
plt.xticks(fontsize=8, rotation=5)
plt.ylabel('Average Salary ($100,000)')
#use the same y-axis scale as in the previous plot for consistent comparison
plt.yticks(ticks=np.arange(0,4.1,0.5), fontsize=9)
ax1.set(ylim=(0,4.2))
plt.grid(axis='y',alpha=0.3,color='gray')
plt.show();
```

The plot above displays the average salary of five jobs in the tech industry. These five jobs have the highest average salaries of jobs that have at least three records in the original dataset. The labels above each bar indicate the number of records each job has. Note that the vertical axis is in units of \\$100,000.

The Head of Machine Learning position stands out slightly in terms of average salary given that it is about \\$26,000 more than that of any other job. Additionally, Heads of Machine Learning see the smallest outright salary of these five positions (\\$76,000) & the most volatility though this latter detail could be attributed to only having nine corresponding records.  
The average salaries of the last three positions are quite comparable in that they only differ by about \\$8,000. The job with the most consistent salaries of these five positions is the Machine Learning Model Engineer despite only having 16 data points.  
The two highest paid jobs in this plot--Head of Machine Learning & Engineering Manager--are also present in the previous plot which contained the five jobs with the overall largest average salaries. Three of these five jobs are associated with machine learning & two of them associate with engineering. Moreover, four of these jobs are positions that oversee the work of others--heads, managers, directors.

# Analysis II - Remote Work & Average Salary
_**Compare the average salaries for employees working remotely 100%, 50%, and 0%. What patterns or trends are present?**_

With continuing developments in technology, the ability for people to work fully remotely & remain effective has become more accessible. Although fully-remote work may impact work within the office building of a company or business, such as ease of communication & working environment, this style of work can also improve accessibility for both companies & employees in terms of who they can hire & where they can work respectively. In hiring employees who work fully remotely, companies can have a wider range of candidates whom they can choose from to fill positions. For hybrid work, employees still need to be near their office & so hiring such candidates will still be as restricted as those who work fully in person.  
Ultimately, whether companies prefer fully-remote or in-person work will depend on various factors, but hybrid work will likely require additional resources & levels of accessibility because companies will have to handle people who work both in the office & from home. As such, hybrid work may be less desirable for businesses.

By evaluating salaries of employees in the tech industry according to how much they worked remotely, it may provide insights as to how prevalent or accessible remote work has become in recent years. If salaries differ, then it may indicate that most companies generally prefer specific types of remote work. As a consequence, employees who may be more restricted to certain types of remote work may face limitations in what salaries they might see. Such information can be relevant to employees &/or companies for which remote work is a factor in the jobs they seek & hire for respectively.

In this dataset, the 'remote_ratio' variable quantifies the percentage of work that an employee does remotely. The survey responses contain three ratios: zero, fifty, & one-hundred percent. To begin, it would be useful to check how many survey responses associate to each of these three ratios to determine a base level of representativeness & credibility of subsequent analyses & findings.

```python
#Analyze salaries across the three 'remote_ratio' values
q2_data_I = salaries_df.groupby('remote_ratio').salary_in_usd.describe().reset_index()
#round salary values to 2 decimals
q2_data_I.iloc[:,2:] = round(q2_data_I.iloc[:,2:],2)
#convert counts to integers
q2_data_I['count'] = q2_data_I['count'].astype('int')
display(q2_data_I)

#visualize the distributions of salaries
ax1 = sns.boxplot(data=salaries_df, x='remote_ratio', y=salaries_df['salary_in_usd']/100000)
#edit plot details
plt.title('Salaries per Remote-Work Type')
plt.xlabel('Remote-Work Type')
ax1.set_xticklabels(labels=['Not Remote', 'Hybrid', 'Fully Remote'], fontsize=9)
plt.yticks(fontsize=9)
plt.ylabel('Salary ($100,000)')
plt.grid(axis='y',alpha=0.1,color='gray')
plt.show();
```

For each of the three 'remote_ratio' values, each has a representative number of data points; however, the number of hybrid-worker ("50%") responses (271) is considerably less than that of the other remote types (12,500+).

The plot above displays the salary distributions of the three remote ratios listed in the 57,194 survey responses of employees within the tech industry. Note that the vertical axis is in units of \$100,000.  
It appears that salaries are more beneficial for employees who commit fully to working either in person or remotely in that they typically earn about \\$150,000. Between these two ratios, the salaries of those who work entirely in person see salaries that are slightly greater than those who work fully remote. Of those who do both (hybrid workers), not only are they significantly fewer in numbers, but they also generally receive lower salaries; about \\$70,000.

```python
#Visualize the avg salary per remote-type
ax1 = sns.barplot(data=q2_data_I, x='remote_ratio', y='mean')
#edit plot details
plt.title('Average Salary per Remote-Work Type')
plt.xlabel('Remote-Work Type')
ax1.set_xticklabels(labels=['Not Remote', 'Hybrid', 'Fully Remote'], fontsize=9)
plt.ylabel('Average Salary ($)')
plt.yticks(fontsize=9)
plt.grid(axis='y',alpha=0.3,color='gray')
plt.show();
```

The plot above displays the average salaries of the three different kinds of remote-work types based on 57,194 survey responses of individuals within the tech industry between 2020-2024.

The patterns in this plot are very similar to those observed in the most previous figure, which displayed the distributions of salaries of these remote-work types. Particularly, the average salary of hybrid workers is significantly less than that of in-person & fully remote workers (by nearly half). Additionally, in-person workers make slightly more than those who work fully remotely on average; by about \\$12,000. 
Recall that the average salary across the whole dataset is about \\$159,000. On average, in-person workers make slightly more than this value, fully remote workers make about \$10,000 less, & hybrid workers make less than half this value.
- People who do no remote work earn roughly \\$160,000 on average.
- People who work evenly between in-person & remote work earn roughly \\$80,000 on average.
- People who work fully remotely earn roughly \\$150,000 on average.

As such, it seems that companies value employees who work in person the most highly. Moreover, employees who don't commit fully to in-person or remote work appear to be valued significantly less than those who do commit to one or the other. It should be noted that there are considerably fewer hybrid records in the dataset than remote or non-remote workers.
