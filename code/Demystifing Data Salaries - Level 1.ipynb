## 💾 The Data

The data, available in `salaries.csv`, comes from a survey hosted by the aforementioned HR consultancy. Each data point represents a single employee's salary record for a given year.

| Column    | Description                                                      |
|-----------|------------------------------------------------------------------|
| `work_year`  | The year the salary was paid. |

| `remote_ratio`  | Percentage of remote work. `0`: No remote work (<20%), `50`: Hybrid (50%), `100`: Fully remote (>80%) |
| `company_location`  | Employer's main office location (ISO 3166 code). |
| `company_size`  | Company size. `S`: Small (<50 employees), `M`: Medium (50–250 employees), `L`: Large (>250 employees) |

# Summary of Findings & Recommendations
This project (part one of three) analyzed over 57,000 survey responses of individuals within the tech industry regarding their employment details between 2020-2024. As no preprocessing was necessary, the analyses examined specific subsets of individuals & their salaries. In particular, employment information of data scientists & data engineers were compared, then ratios & salaries of people based in the US who work full-time & fully remotely were evaluated. In assessing such individuals, a deeper understanding as to some recent/current trends of employees & salaries within the tech industry was uncovered. Such information can be used to assist companies within the industry in attracting the best candidates & to help drive success.

### Analysis I - Number of Annual Records
The HR consultancy obtained 57,194 survey responses from individuals across the globe between 2020-2024. More responses were obtained each year. In particular, 2024 accounted for almost 82% of the survey responses & 2023-2024 accounted for around 95% of them. There were only 75 responses in the first year that this survey was conducted (2020). The bias in these data points towards more recent years demonstrates the relevancy of much of this data therefore making it more reliable.

### Analysis II - Salaries of Data Scientists, Data Engineers
Of the 57,194 survey responses, 17,676 were from data scientists & data engineers. Over the five years of records, the average salary of data scientists (~ \\$159,000) was about \\$10,000 more than that of data engineers (~ \\$149,000), the former of which is roughly equivalent to the overall average salary across the 57,194 responses.  
When evaluated on a yearly basis, both positions saw an overall increase in their average salary between 2020 & 2024. That being said, the annual average salary of data engineers increased incrementally each year, whereas that of data scientists decreased from 2020 to 2021 & 2023 to 2024. This might be indicative that data engineer salaries are slightly less volatile & more predictable than those of data scientists. Regardless, the average salary of data scientists was greater than that of data engineers in four of the five years of records.

Based on these trends, rough estimates can be generated regarding the potential future salaries of these individuals. While typical salaries of both positions should be expected to continue to grow in the coming years, less growth in the last three years (2022-2024) might put salary estimates in 2030 roughly around \\$180,000 & \\$220,000 for data engineers & data scientists respectively. In particular, the bumpier trends in reported salaries of data scientists will make forecasting their salaries more difficult.


### Analysis III - Full-time Remote Workers
Of the 57,194 responses, 11,125 were from employees who worked fully remote, full-time, & were US-based between 2020-2024. Over this period, the average salary of such individuals was about \\$155,000, slightly less than that of the overall average reported salary (~ \\$159,000).  
When evaluated on a yearly basis, the recorded average salaries of these individuals were relatively consistent as there was only a ~\\$1,800-overall increase from 2020 to 2024. Apart from a slight dip of about \\$12,000 from 2020 to 2021, the average annual salaries were between \\$151,000-\\$156,000 for these employees across four of these five years. Moreover, this average only changed by about -\\$240 in the most recent two years. Such consistency could help provide a competitive, reliable baseline for companies looking to hire such individuals.

Based on these trends, rough estimates can be generated regarding the potential future salaries of these individuals. Given the very small growth rate observed across the reported salaries of these individuals, their salaries in 2030 would be projected at around about \\$166,000.

Furthermore, the proportions of US-based, full-time employees who worked fully remotely were evaluated over the five-year period to see how the rates evolved. Over the first three years, the rate of these individuals who worked entirely remotely increased--more than two people of every five worked fully remotely in 2022--but this rate declined precipitously from 2022-2024. It was hypothesized that these trends were related to the COVID-19 pandemic (which was most pressing in 2020) given that more people were forced to work from home but were eventually able to return back to "normalcy." Alternatively, such trends could point to employees &/or companies having less desire for fully-remote work. Regardless, the most recent trend points to fully-remote work becoming less popular among these individuals.

# Analysis I - Number of Annual Records
_**How many records are in the dataset, and what is the range of years covered?**_

As discovered previously, there are 57,194 salary records in the provided dataset which span 2020-2024. Additional analysis can be done to see how recent these data points are thus giving an idea as to how relevant they might be.

```python
#Explore the 'work_year' variable
#count # records per year
records_per_year = salaries_df['work_year'].value_counts().sort_index().reset_index()
#add a column indicating the proportion of records per year
records_per_year['percentage'] = round((records_per_year['count']/salaries_df.shape[0])*100, 2)
display(records_per_year)

#visualize the data
ax1 = sns.countplot(data=salaries_df, x='work_year')
#label each bar with the % of total records
ax1.bar_label(ax1.containers[-1], labels=[f"{num} %" for num in records_per_year['percentage']], fontsize=9)
#edit plot details
plt.title('# Records per Year')
plt.xlabel('Year')
plt.ylabel('# Records')
plt.yticks(ticks=range(0,50001,10000), fontsize=9)
plt.grid(axis='y',alpha=0.1,color='gray')
plt.show();
```

The plot above displays the number of records for each unique year in which an individual reported to their salary. The data, comprised of 57,194 records, spans 2020-2024. The percentages above each bar indicate how many observations of the entire dataset each year accounts for.

From this plot, the number of records appears to increase exponentially on a yearly basis. As such, this suggests that the survey that supplies the dataset compiled exponentially more employee salary records over these five years. Alternatively, the HR consultancy that hosted this survey may have obtained access to many more individuals over time who were willing to answer their survey. Nonetheless, the significant number of records from recent years, particularly 2024, demonstrates the relevancy of much this data.  
On another note, it would be interesting to evaluate the impacts of the COVID-19 pandemic, which was primarily relevant in 2020, & whether it may have had consequences on this dataset & the HR consultancy's ability to conduct this survey &/or obtain responses.

# Analysis II - Salaries of Data Scientists, Data Engineers
_**What is the average salary (in USD) for Data Scientists and Data Engineers? Which role earns more on average?**_

By evaluating average salaries of data scientists & data engineers, it can provide a stronger understanding as to how their salaries have generally evolved & what companies should expect to spend & offer when attracting potential candidates for such positions. This knowledge can help such companies stay relevant, competitive, & appealing for these kinds of candidates/employees.

To start, gaining an initial understanding of all individuals' salaries within the dataset can help assess whether data scientists & data engineers typicaly earn more or not.

```python
#Explore the different jobs in the dataset
num_jobs = salaries_df.job_title.value_counts().reset_index()
num_jobs['percentage'] = num_jobs['count']/salaries_df.shape[0]
display(num_jobs.head(5))

#Explore employee salaries (USD)
display(salaries_df.salary_in_usd.describe().reset_index().T)
```

Of the 57,194 responses in this survey, data scientists & data engineers are the two-most popular positions--they account for 17,767 (~ 30%) of the responses.  
Overall, reported salaries are on average about \\$159,000 but they range from \\$15,000-\\$800,000. With this in mind, the salaries of these two specific positions can be examined.

```python
#Filter for the jobs of interest
data_sci_eng = salaries_df[salaries_df.job_title.isin(['Data Scientist', 'Data Engineer'])]
#need to categorize the "job_title" variable to help make plotting consistent
data_sci_eng['job_title'] = pd.Categorical(data_sci_eng['job_title'], categories=['Data Scientist','Data Engineer'], 
                                           ordered=True)

#Calculate salary metrics for these jobs
data_sci_eng_stats = data_sci_eng.groupby('job_title')['salary_in_usd'].describe()

#reorganize the data
data_sci_eng_stats = data_sci_eng_stats.T.reset_index().melt(id_vars='index', 
                                                             value_vars=['Data Engineer','Data Scientist']).rename(
                        columns={'index':'metric', 'value':'salary_in_usd', 'level_0':'index'})
#round the salaries
data_sci_eng_stats['salary_in_usd'] = round(data_sci_eng_stats['salary_in_usd'], 2)

display(data_sci_eng_stats.sort_values('metric').reset_index(drop=True).iloc[6:12,:])
```

```python
#Visualize the data
ax1 = sns.boxplot(data=data_sci_eng, x='job_title', y=data_sci_eng['salary_in_usd']/100000, fill=False)
#add the mean salary
sns.scatterplot(data=data_sci_eng_stats[data_sci_eng_stats.metric == 'mean'], x='job_title', 
                y=data_sci_eng_stats['salary_in_usd']/100000, marker='*', color='red', label="Mean", s=120)

#edit plot details
plt.title('Salaries of Data Scientists & Data Engineers')
plt.xlabel('Job')
plt.xticks(fontsize=9)
plt.ylabel('Salary ($100,000)')
plt.yticks(ticks=range(0,9,1), fontsize=9)
plt.legend(fontsize=9, loc='upper right')
plt.grid(axis='y',alpha=0.1,color='gray')
plt.show();
```

The figure above illustrates the distribution of salaries of data scientists & data engineers (in USD) of those who responded to the survey between 2020-2024. Note that the vertical axis is in units of \$100,000. The red stars denote the average salary for each position.

Overall, it appears that average salaries of data scientists are generally slightly greater than those of data engineers. Specifically, the average salary of data scientists is about \\$10,000 more than that of data engineers.  
Recall that across all jobs in the dataset, the typical average salary is about \$159,000. As such, data scientists typically make about this amount based on this survey, whereas data engineers make slightly less (by about \\$10,000).

```python
#Analyze the salaries of these positions over time
#count # records per year
data_sci_eng_nums = data_sci_eng.groupby(['work_year','job_title']).size().reset_index().rename(columns={0:'num'})
#display(data_sci_eng_nums)

#Retrieve mean annual salaries
data_sci_eng_yr_avg_sal = data_sci_eng.groupby(['work_year','job_title'])['salary_in_usd'].mean().reset_index().rename(
                            columns={'salary_in_usd':'avg_salary'})
display(data_sci_eng_yr_avg_sal)
```

```python
#Visualize the data
fig, axes = plt.subplots(nrows=2, figsize=(9,9))
ax1 = sns.lineplot(data=data_sci_eng_yr_avg_sal, x='work_year', y=data_sci_eng_yr_avg_sal['avg_salary']/100000, 
                   hue='job_title', marker='o', ax=axes[0])
ax2 = sns.boxplot(data=data_sci_eng, x='work_year', y=data_sci_eng['salary_in_usd']/100000, hue='job_title', ax=axes[1])

#edit plot details
fig.suptitle("Reported Annual Salaries of Data Scientists & Data Engineers", fontsize=12)
for k,v in {"Average Annual Salaries":[0, 'Average Salary ($100,000)', 0,2.1,0.25],
           "Annual Salaries":[1, 'Salary ($100,000)', 0,8,1]}.items():
    plt.sca(axes[v[0]])
    plt.title(k)
    plt.xlabel("Year")
    #for the line plot
    if v[0]==0:
        plt.xticks(ticks=range(2020,2025,1), fontsize=9)
    plt.xticks(fontsize=9)
    plt.ylabel(v[1])
    plt.yticks(fontsize=9, ticks=np.arange(v[2],v[3],v[4]))
    plt.legend(title="Job", fontsize=9, loc='upper left')
    plt.grid(axis='y',alpha=0.1,color='gray')
    plt.tight_layout();
```

The plots above illustrate the annual evolution in the reported salaries of data scientists & data engineers between 2020-2024. Of the 57,194 responses in the survey, 17,676 of them pertain to these two positions. The first (line) plot shows the average annual salaries of these positions. Note that the vertical axes, which are in units of \\$100,000, are not identical across the two plots.

In the first plot, there is a general linear trend in the average annual salaries of data engineers from 2020-2024; however, the annual average salary of data scientists decrease first from 2020 to 2021 & then again from 2023 to 2024 which suggests that salaries of data engineers are slightly less volatile & more predictable. In four of the five years, data scientists had a larger average salary than data engineers. In the last two years, this difference was between about \\$10,000-\\$15,000 in both 2023 & 2024.  
Overall, both positions have seen a considerable increase in salary across this period. From 2020 to 2024, the average salaries of data engineers & data scientists grew by approximately \\$65,000 & \\$51,000 respectively. While typical salaries of both positions should be expected to continue to grow in the coming years, less growth in the last three years (2022-2024) might put salary estimates in 2030 roughly around \\$180,000 & \\$220,000 for data engineers & data scientists respectively. In particular, the bumpier trends in reported salaries of data scientists will make forecasting their salaries more difficult.

In the second plot, similar trends are evident from 2020-2023, the main difference being that it is more of a positive trend for data scientists from 2020-2023. On another note, reported salaries greater than \\$650,000 were all in 2023 or 2024. Given that the number of survey responses increased so significantly in the last two years compared to the first three, a larger sample size could have contributed to observing these massive salaries. This seems more likely than a major salary increase from 2022 to 2023 onwards--the largest salary reported before 2023 was around \\$415,000 (2020) which is considerably smaller than the most significant salaries reported in 2023-2024 of more than \\$650,000.

# Analysis III - Full-time Remote Workers
_**How many full-time employees based in the US work 100% remotely?**_

In recent years, as technology has advanced & other events have influenced peoples' daily ways of life (i.e. the COVID-19 pandemic), it has impacted the ways in which employees work. More specifically, these factors & others have contributed to people working outside of the office more & more. By analyzing such information of individuals, it can provide an assessment as to how peoples' working environments have evolved recently as well as how much of a priority it is becoming for candidates regarding their desire to work remotely. If working remotely is becoming more desirable for employees, then companies, speciically based in the US, may need to consider this factor more heavily when searching for & hiring potential candidates.

To obtain records of fully remote, full-time employees based in the US, the data needs to be filtered appropriately. Beforehand, exploring the relevant variables might be useful to gain an initial understanding of them.
- For clarity, when identifying records of individuals "based in the US," the 'employee_residence' variable should be used given that it relates to an employee's primary country of residence.

```python
#Explore the three variables of relevance
#Employment_type
num_emp_type = salaries_df.groupby('employment_type').size().reset_index().rename(columns={0:'num'})
num_emp_type['percentage'] = round(100*num_emp_type['num']/salaries_df.shape[0],2)
display(num_emp_type)
#Employee_residence
num_emp_res = salaries_df.groupby('employee_residence').size().reset_index().rename(columns={0:'num'})
num_emp_res['percentage'] = round(100*num_emp_res['num']/salaries_df.shape[0],2)
display(num_emp_res[num_emp_res.employee_residence == "US"])
#Remote_ratio
num_remote = salaries_df.groupby('remote_ratio').size().reset_index().rename(columns={0:'num'})
num_remote['percentage'] = round(100*num_remote['num']/salaries_df.shape[0],2)
display(num_remote)
```

Based on the data above, some notes regarding the relevant variables can be pointed out.
- About 99.6% of employees in the dataset work full-time.
- About 90.7% of employees in the dataset are primarily based in the US.
- About 22.0% of employees in the dataset work entirely remotely, which is the second most popular segment.

```python
#Filter for full-time, fully remote employees based in the US
data_III = salaries_df[(salaries_df.employment_type == "FT") & (salaries_df.employee_residence == "US") & (
                salaries_df.remote_ratio == 100)]
#List the number of relevant records
display("Of the 57,194 records in the dataset, {} of them pertain to employees who work full-time, are based in the US, & work completely remotely; about {}%.".format(
                data_III.shape[0], round(100*data_III.shape[0]/salaries_df.shape[0], 1)))

#Analyze these salaries
display(round(data_III['salary_in_usd'].describe(), 2).reset_index().T)
```
